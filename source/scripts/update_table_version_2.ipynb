```python
import pandas as pd
```

```python
# Load table ../../data/games.csv with pandas
df = pd.read_csv('../../data/games.csv')
```

```python
df.tail()
```

```python
# Clean the dataframe where rows with the column "name" are empty
df = df[df['name'].notna() & (df['name'] != '')]
```

```python
df.tail()
```

```python
# Clean the dataframe where rows with the column "name" must not repeat
df = df.drop_duplicates(subset=['name'], keep='first')
```

```python
df.tail()
```

```python
# Save the new dataframe with name ../../data/games_2.csv
df.to_csv('../../data/games_2.csv', index=False)
print("Data cleaning complete. Saved to ../../data/games_2.csv")
```
