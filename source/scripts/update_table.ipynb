```python
import pandas as pd
```

```python
# Load table ../../data/games.csv with pandas
df = pd.read_csv('../../data/games.csv')
```

```python
df.head()
```

```python
# Clean the dataframe where the column "id" is removed
df = df.drop(columns=['id'], errors='ignore')
```

```python
df.tail()
```

```python
# Clean the dataframe where rows with the column "name" are empty
df = df[df['name'].notna() & (df['name'] != '')]
```

```python
df.tail()
```

```python
# Clean the dataframe where rows with the column "name" must not repeat
df = df.drop_duplicates(subset=['name'], keep='first')
```

```python
df.tail()
```

```python
# Load gamespot reviews
df_reviews = pd.read_csv("../../data/reviews.csv")
unique_games = [f"{game_name}" for game_name in df_reviews["game_name"].unique()]
unique_games.sort()
```

```python
unique_games[::-1]
```

```python
from thefuzz import fuzz

# Function to decide the equalness of two strings
def nameRatio(name1: str, name2: str) -> bool:
    return fuzz.ratio(name1, name2) / 100.0

nameRatio("RealMyst", "RealMyst 2")
```

```python
from tqdm import tqdm

map_game_names = {}
for game in tqdm(df["name"], desc="Mapping game names"):
    # Find best match
    max_name_ratio = 0.0
    max_index = 0
    for index, game_name in enumerate(unique_games):
        current_name_ratio = nameRatio(game, game_name)
        if current_name_ratio > max_name_ratio:
            max_name_ratio = current_name_ratio
            max_index = index
    # Map result of query game names to name in unique games
    map_game_names[game] = unique_games[max_index]
```

```python
map_game_names["Red Dead Redemption"]
```

```python
# Create a new column "real_name" that holds the mapped game names
df["real_name"] = df["name"].map(map_game_names)
```

```python
# Reorder columns to place "real_name" first
cols = df.columns.tolist()
cols.remove('real_name')
df = df[['real_name'] + cols]
```

```python
# Save the new dataframe with name ../../data/games_2.csv
df.to_csv('../../data/games_2.csv', index=False)
print("Data cleaning complete. Saved to ../../data/games_2.csv")
```
